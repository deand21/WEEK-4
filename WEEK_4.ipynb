```python
# LATIHAN 1
# Program Menentukan Tahun Kabisat

tahun = int(input("Masukkan tahun: "))

kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print("Tahun kabisat:", kabisat)
```

```python
# LATIHAN 2
# Program Menghitung Energi Listrik
# Input  : Daya dan waktu
# Output : Energi listrik

def hitung_energi(daya, waktu):
    energi = daya * waktu
    return energi


# Input
daya = float(input("Masukkan daya listrik (Watt): "))
waktu = float(input("Masukkan waktu penggunaan (jam): "))

# Proses
energi = hitung_energi(daya, waktu)

# Output
print("Energi listrik yang digunakan =", energi, "Wh")
```
